```python
#TRIM and FIX spacing 
```

```python
df["Name"] = df["Name"].str.strip()
```

```python
# ---- 2. PROPER case fix (Find & Replace / PROPER equivalent) ----
```

```python
df["Name"] = df["Name"].str.title()
df["payment_mode"] = df["payment_mode"].str.strip().str.title()
df["order_status"] = df["order_status"].str.strip().str.title()
```

```python
#Standardize gender
#notna : "Is this value NOT missing?" 
#.map() applies a function to every value in the column.
# gender_map.get(x, x) This is dictionary lookup.
```

```python
gender_map = {
    "M":"Male","MALE":"Male", "Male": "Male", "m": "Male",
    "F": "Female", "FEMALE": "Female", "Female": "Female", "f": "Female"
}
df["gender"] = df["gender"].str.strip().str.upper().map(
    lambda x:gender_map.get(x,x) if pd.notna(x) else x )
```

```python
# safer version using your original casing as fallback:
df["gender"] = df["gender"].replace({
    "M": "Male", "MALE": "Male", "m": "Male",
    "F": "Female", "FEMALE": "Female", "f": "Female"
})
    
```

```python
# ---- 4. Flag bad data (Conditional Formatting equivalent) ----
```

```python
df["flag_negative_qty"] = df["quantity"] < 0
df["flag_invalid_discount"] = df["quantity"] > 1
print("Negative quantity rows:", df["flag_negative_qty"].sum())
print("Invalid discount rows:", df["flag_invalid_discount"].sum())
```

```python
# ---- 5. Remove exact duplicate rows ----
```

```python
len(df)
```

```python
before = len(df)
df = df.drop_duplicates()
print(f"Removed {before - len(df)} exact duplicates rows")
```

```python
# Flag duplicate order_id WITHOUT deleting (same as your COUNTIF formula)
```

```python
df["duplicate_order_id"] = df["order_id"].duplicated(keep=False).map(
    {True: "duplicates", False: "unique"}
)
print(df["duplicate_order_id"].value_counts())
```

```python
# ---- 6. Pivot Tables (Revenue by Category, Revenue by Region) ----
```

```python
pivot_category = df.pivot_table(values="revenue", index="product_category", aggfunc="sum").round(2)
pivot_region = df.pivot_table(values="revenue", index="region", aggfunc="sum").round(2)
```

```python
print(pivot_category)
print(pivot_region)
```

```python
# a. Rename columns for consistenc
```

```python
df = df.rename(columns={
    "Name" : "customer_name",
    "gender" : "customer_gender",
    "DOB" : "date_of_birth"
})
```

```python
# b. Drop irrelevant columns
```

```python
df= df.drop(columns = ["interest"])
```

```python
# c. Fix data types
```

```python
df["date_of_birth"] = pd.to_datetime(df["date_of_birth"], errors="coerce")  # handles mixed formats
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
```

```python
# d. Standardize categorical values
```

```python
df["customer_name"] = df["customer_name"].str.strip().str.title()
df["customer_gender"] = df["customer_gender"].str.upper().map({
    "M": "Male", "MALE": "Male", "F": "Female", "FEMALE": "Female"
}).fillna(df["customer_gender"])
df["country"] = df["country"].str.strip().str.upper().replace({
    "INDIA": "India", "USA": "United States", "U.S.A": "United States",
    "UK": "United Kingdom"
})
df["payment_mode"] = df["payment_mode"].str.strip().str.title().replace({"Cc": "Credit Card", "Cod": "Cash On Delivery"})
df["order_status"] = df["order_status"].str.strip().str.title()
```

```python
# e. Fix missing values using existing data
```

```python
df["country"] = df["country"].fillna(df["country"].mode()[0])
mode_by_category = df.groupby("product_category")["customer_segment"].transform(lambda x: x.mode()[0] if not x.mode().empty else "Consumer")
df["customer_segment"] = df["customer_segment"].fillna(mode_by_category)
```

```python
# f. Fix invalid business-logic values
```

```python
# f. Fix invalid business-logic values
df = df[df["quantity"] > 0]                     # remove impossible negative quantities
df["discount"] = df["discount"].clip(0, 0.5)     # cap discount between 0% and 50%
```

```python
# g. Recompute revenue instead of trusting missing/wrong values
```

```python
df["revenue"] = (df["quantity"] * df["unit_price"] * (1 - df["discount"])).round(2)
```

```python
# h. Remove exact duplicate rows
```

```python
df = df.drop_duplicates()
```

```python
# i. Handle duplicate order_ids (keep latest/highest revenue entry as business rule)
```

```python
df = df.sort_values("revenue", ascending=False).drop_duplicates(subset="order_id", keep="first")

df.to_csv("ecommerce_sales_clean.csv", index=False)
```
